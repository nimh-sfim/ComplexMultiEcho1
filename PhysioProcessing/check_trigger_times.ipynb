```python
# Quick script to check trigger times to confirm the TR
import os.path as op
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from glob import glob

```

```python
base_dir = '/Volumes/NIMH_SFIM/handwerkerd/ComplexMultiEcho1/Data'
sampling_rate = 2000 #Hz
print(op.join(base_dir, 'sub-*', 'Unprocessed/*_physio.tsv.gz'))
physio_files = glob(op.join(base_dir, 'sub-*', 'Unprocessed/func/*_physio.tsv.gz'))
print(physio_files)
```

```python
trigger_times_mean_std_max_min = np.full((len(physio_files), 4), np.nan)
for file_idx, physio_file in enumerate(physio_files):
    physio_data = pd.read_csv(physio_file, sep='\t', compression='gzip')
    trigger_positive = np.where(physio_data['Trigger'] > 1)[0]
    trigger_diffs = np.diff(trigger_positive)
    trigger_times = trigger_diffs[trigger_diffs > 1]/sampling_rate

    trigger_times_mean_std_max_min[file_idx, 0] = np.mean(trigger_times)
    trigger_times_mean_std_max_min[file_idx, 1] = np.std(trigger_times)
    trigger_times_mean_std_max_min[file_idx, 2] = np.max(trigger_times)
    trigger_times_mean_std_max_min[file_idx, 3] = np.min(trigger_times)

    #print(f"{file_idx=}, {trigger_times_mean_std_max_min[file_idx, :]=}")
```

```python
print("num files:", len(physio_files))
print("means (ms):", np.round(trigger_times_mean_std_max_min[:, 0],5)*1000)
print("means (ms diff from 1500ms):", np.round(1.5-trigger_times_mean_std_max_min[:, 0],5)*1000)
print("stds (ms):", np.round(trigger_times_mean_std_max_min[:, 1],5)*1000)
print("maxs (ms diff from 1500ms):", np.round(1.5-trigger_times_mean_std_max_min[:, 2],5)*1000)
print("mins (ms diff from 1500ms):", np.round(1.5-trigger_times_mean_std_max_min[:, 3],5)*1000)
```
